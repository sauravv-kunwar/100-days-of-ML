# ElasticNet Regression

## Definition
ElasticNet Regression is a type of linear regression that combines **Ridge Regression (L2)** and **Lasso Regression (L1)**.

It is used to reduce overfitting and also perform feature selection.

---

# Why ElasticNet?

- Ridge reduces coefficient size but does not remove features
- Lasso removes features but can be unstable when features are correlated
- ElasticNet combines both advantages

---

# Formula

## Linear Regression Model

$$
\hat{y} = b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n
$$

---

## ElasticNet Cost Function

$$
J(\beta) = \sum (y_i - \hat{y}_i)^2 + \lambda_1 \sum |\beta_j| + \lambda_2 \sum \beta_j^2
$$

OR

$$
J(\beta) = RSS + \lambda_1 ||\beta||_1 + \lambda_2 ||\beta||_2^2
$$

Where:

- $RSS$ = Residual Sum of Squares
- $\lambda_1$ = L1 regularization strength (Lasso)
- $\lambda_2$ = L2 regularization strength (Ridge)

---

# Key Idea

ElasticNet applies both:

- L1 penalty → Feature selection (like Lasso)
- L2 penalty → Stability (like Ridge)

---

# Important Parameter

### l1_ratio

Controls balance between Lasso and Ridge:

- l1_ratio = 1 → Pure Lasso
- l1_ratio = 0 → Pure Ridge
- 0 < l1_ratio < 1 → ElasticNet

---

# Advantages

- Handles correlated features better than Lasso
- Performs feature selection
- Reduces overfitting
- More stable than Lasso

---

# Disadvantages

- More hyperparameters to tune
- Slightly more complex than Ridge or Lasso

---

# When to Use ElasticNet?

Use ElasticNet when:

- Dataset has many features
- Features are correlated
- You want both regularization + feature selection

---


```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Ridge 
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```
