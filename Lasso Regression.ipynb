# Lasso Regression

## Definition
Lasso Regression is a type of Linear Regression that uses **L1 Regularization** to reduce overfitting.

It can also remove unnecessary features by making some coefficients equal to zero.

---

# Formula

## Linear Regression Formula

$$
\hat{y} = b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n
$$

---

## Lasso Regression Cost Function

$$
J(\beta) = \sum (y_i - \hat{y}_i)^2 + \lambda \sum |\beta_j|
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $\beta_j$ = Coefficients
- $\lambda$ = Regularization parameter

---

# Main Idea

Lasso adds a penalty to large coefficients.

- Small $\lambda$ → Less regularization
- Large $\lambda$ → More regularization

Some coefficients become:

$$
\beta = 0
$$

So unimportant features are removed.

---

# Advantages

- Reduces overfitting
- Performs feature selection
- Makes model simpler

---

# Disadvantages

- May remove important features
- Not stable for highly correlated data

---

# Difference Between Ridge and Lasso

| Ridge Regression | Lasso Regression |
|---|---|
| Uses L2 Regularization | Uses L1 Regularization |
| Does not remove features | Removes features |
| Penalty: $\sum \beta^2$ | Penalty: $\sum |\beta|$ |

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)

plt.scatter(X,y)

from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X_train,y_train)
print(reg.coef_)
print(reg.intercept_)
```

```python
alphas = [0,1,5,10,30]
plt.figure(figsize=(12,6))
plt.scatter(X,y)
for i in alphas:
    L = Lasso(alpha=i)
    L.fit(X_train,y_train)
    plt.plot(X_test,L.predict(X_test),label='alpha={}'.format(i))
plt.legend()
plt.show()
```

```python
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

def get_preds_lasso(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('lasso', Lasso(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 0.1, 1]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_lasso(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```
