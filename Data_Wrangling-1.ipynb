# Data Wrangling
- Data wrangling (also known as data munging or data cleaning) is the process of transforming raw, messy, or unstructured data into a clean, structured, and usable format suitable for analysis, visualization, or modeling.
- Simple Definition
- Data wrangling = Getting data from its “real-world” messy state into a tidy, analysis-ready state.
# Why Data Wranling Matters??
- In real-world data science projects:
- 80% of the time is spent on data wrangling
- Only 20% on actual modeling/visualization
- Raw data is rarely perfect — it often has:
- Missing values
- Inconsistent formats (e.g., dates as strings)
- Duplicates
- Wrong data types
- Outliers
- Mixed case text
- Unnecessary columns
# Key Steps in Data Wrangling
### Gathering/Ingestion
- Load data from files (CSV, Excel), databases, APIs, web pages
### Assessing
- Explore: df.head(), df.info(), df.describe(), check for missing/duplicates
### Cleaning
- Handle missing values (drop/fill/impute)
- Remove duplicates
- Fix incorrect values
### Transforming
- Rename columns
- Change data types (e.g., string → datetime)
 - Create new features (e.g., age from birth year)


## Hierarchical Indexing (MultiIndex) in Pandas

Hierarchical indexing, also known as **MultiIndex**, allows pandas objects (Series or DataFrame) to have **multiple levels of indexing on one axis**. It enables representation of higher-dimensional data in a lower-dimensional structure.

With hierarchical indexing, each data point is identified by **more than one index label**, creating a hierarchy of indexes.

### Key Features
- Allows **multiple index levels** for rows or columns.
- Helps organize **complex datasets with grouped labels**.
- Supports powerful **data reshaping operations** such as `stack()` and `unstack()`.

### Example

```python
import pandas as pd

data = pd.Series(
    [10, 20, 30, 40],
    index=[['A', 'A', 'B', 'B'], [1, 2, 1, 2]]
)

# Merging and Combining Datasets
#### Ways of combining pandas objects
### pandas.merge

- Connect rows in DataFrames based on one or more keys. This will be familiar to users of SQL or other relational databases, as it implements database join operations.

### pandas.concat

- Concatenate or “stack” objects together along an axis.

### combine_first

- Splice together overlapping data to fill in missing values in one object with values from another.

````python
df1 = pd.DataFrame({"key": ["b", "b", "a", "c", "a", "a", "b"],
                    "data1": pd.Series(range(7), dtype="Int64")})
df2 = pd.DataFrame({"key": ["a", "b", "d"],
                    "data2": pd.Series(range(3), dtype="Int64")})
df1
````

````python
df2
````

````python
pd.merge(df1,df2) # performs the join through common key,only the common keys are left others are ignored
````

````python
pd.merge(df1,df2,on = 'key')  # Inner join is performed defaultly when merge operation is performed
````

````python
df3 = pd.DataFrame({"lkey": ["b", "b", "a", "c", "a", "a", "b"],
                    "data1": pd.Series(range(7), dtype="Int64")})
df4 = pd.DataFrame({"rkey": ["a", "b", "d"],
                    "data2": pd.Series(range(3), dtype="Int64")})
pd.merge(df3, df4, left_on="lkey", right_on="rkey")
````

````python
# Outer Merge---> it keeps all the keysb from both DF ,filling missing values with NaN
pd.merge(df1, df2, how="outer")
pd.merge(df3, df4, left_on="lkey", right_on="rkey", how="outer")
````

````python
# how = 'inner'--> use only the key combo observed in both tables
# how  = 'left' ---> use all key combination found in the left tables
# how = 'right' --> use all key combination found in right table 
# how = 'outer'  --> use all key combination observed in both tables together
````

````python
df1 = pd.DataFrame({"key": ["b", "b", "a", "c", "a", "b"],
                    "data1": pd.Series(range(6), dtype="Int64")})
df2 = pd.DataFrame({"key": ["a", "b", "a", "b", "d"],
                    "data2": pd.Series(range(5), dtype="Int64")})
df1
````

````python
df2
````

````python
pd.merge(df1,df2,on = 'key', how = 'left' )  # only keep keys from the left side and fill the value nan for the data in right side
````

````python
pd.merge(df1,df2, on = 'key', how = 'right')
````

````python
pd.merge(df1,df2, on = 'key', how = 'outer')
````

# Reshaping and Pivoting
#### stack
- This “rotates” or pivots from the columns in the data to the rows.
### unstack
- This pivots from the rows into the columns.

````python
data = pd.DataFrame(np.arange(6).reshape((2, 3)),
                    index=pd.Index(["Ohio", "Colorado"], name="state"),
                    columns=pd.Index(["one", "two", "three"],
                    name="number"))
data
````

````python
result = data.stack()
result
````

````python
result.unstack()
````

````python
result.unstack(level=0)
result.unstack(level="state")
````

````python
s1 = pd.Series([0, 1, 2, 3], index=["a", "b", "c", "d"], dtype="Int64")
s2 = pd.Series([4, 5, 6], index=["c", "d", "e"], dtype="Int64")
data2 = pd.concat([s1, s2], keys=["one", "two"])
data2
````

````python
data2.unstack()
````

````python
data2.unstack().stack()
````

````python
data2.unstack().stack(dropna = False)
````

````python
df = pd.DataFrame({"left": result, "right": result + 5},
                  columns=pd.Index(["left", "right"], name="side"))
df
df.unstack(level="state")
````

````python
df.unstack(level="state").stack(level="side")
````

````python
# data = pd.read_csv("data/macrodata.csv")
# data = data.loc[:, ["year", "quarter", "realgdp", "infl", "unemp"]]
# data.head()
````

````python
df = pd.DataFrame({"key": ["foo", "bar", "baz"],
                   "A": [1, 2, 3],
                   "B": [4, 5, 6],
                   "C": [7, 8, 9]})
df
````

````python
melted = pd.melt(df,id_vars = 'key')
melted

#pd.melt reshapes the data from wide to long format
# id_vars='key' → keep this column as it is (don’t melt it)

# All other columns (A, B, C) → melted into two columns:

# variable → the original column name (A/B/C)

# value → the value in that column
````

````python
reshaped = melted.pivot(index="key", columns="variable",
                        values="value")
reshaped
````

````python
reshaped.reset_index()
````

````python
pd.melt(df, id_vars="key", value_vars=["A", "B"])
# id_vars = 'key' --> keep this column as in
# 
````

````python
pd.melt(df, value_vars=["A", "B", "C"])
pd.melt(df, value_vars=["key", "A", "B"])
````
