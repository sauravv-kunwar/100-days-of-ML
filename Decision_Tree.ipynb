## Decision Trees

A Decision Tree is a non-parametric supervised learning algorithm used for classification and regression tasks. It works by creating a model that predicts the value of a target variable by learning simple decision rules inferred from the data features. The structure resembles a flowchart, where each internal node represents a 'test' on an attribute, each branch represents the outcome of the test, and each leaf node represents a class label (in classification) or a numerical value (in regression).

### Key Concepts:

*   **Root Node**: The top-most node in the tree, representing the entire dataset.
*   **Internal Node**: A node that splits into two or more branches, representing a decision based on a feature.
*   **Leaf Node (Terminal Node)**: A node that does not split further, representing the final decision or prediction.
*   **Splitting**: The process of dividing a node into two or more sub-nodes.

### How it Works (Briefly):

*   The algorithm recursively splits the dataset into subsets based on the values of the attributes.
*   Splitting criteria are chosen to maximize the purity of the child nodes.
*   Common impurity measures include:

    #### Gini Impurity (for classification):

    $G = 1 - \sum_{i=1}^{c} (p_i)^2$

    *   Where $p_i$ is the proportion of observations labeled with class $i$ in the node.
    *   $c$ is the number of classes.

    #### Entropy (for classification):

    $E = - \sum_{i=1}^{c} p_i \log_2(p_i)$

    #### Information Gain (used with Entropy):

    $IG(S, A) = E(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} E(S_v)$

    *   Where $S$ is the set of examples.
    *   $A$ is an attribute.
    *   $Values(A)$ are the possible values for attribute $A$.
    *   $|S_v|$ is the number of examples in subset $S_v$.
    *   $|S|$ is the total number of examples.

    #### Mean Squared Error (MSE) (for regression):

    $MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$

    *   Where $n$ is the number of samples.
    *   $y_i$ is the actual value.
    *   $\hat{y}_i$ is the predicted value.

### Advantages:

*   Easy to understand and interpret.
*   Requires little data preparation.
*   Can handle both numerical and categorical data.

### Disadvantages:

*   Prone to overfitting.
*   Can be unstable (small variations in data can lead to a completely different tree).

## Decision Trees: Geometric Intuition, Entropy, Gini Impurity, and Information Gain

### Geometric Intuition

*   Decision Trees work by recursively partitioning the feature space into a set of rectangular regions.
*   Each split in the tree corresponds to a hyperplane (a line in 2D, a plane in 3D, etc.) that is parallel to one of the axes.
*   The goal is to create regions where the data points within each region are as 'pure' as possible, meaning they predominantly belong to a single class (for classification) or have similar target values (for regression).
*   For example, in a 2D space with features X1 and X2, the tree might first split based on `X1 < threshold1`, then `X2 < threshold2`, creating rectangular decision boundaries.

### Entropy

*   **Definition**: Entropy is a measure of the disorder or impurity within a set of data points.
*   **Formula (for classification)**: $E = - \sum_{i=1}^{c} p_i \log_2(p_i)$
    *   $p_i$: Proportion of samples belonging to class $i$ in the current node.
    *   $c$: Number of classes.
*   **Role in Decision Trees**: The algorithm seeks to minimize entropy in the child nodes after a split. A node with low entropy is considered 'pure' because most of its samples belong to one class.

### Gini Impurity

*   **Definition**: Gini impurity is another measure of impurity in a set of data. It quantifies the probability of incorrectly classifying a randomly chosen element in the dataset if it were randomly labeled according to the distribution of labels in the subset.
*   **Formula (for classification)**: $G = 1 - \sum_{i=1}^{c} (p_i)^2$
    *   $p_i$: Proportion of samples belonging to class $i$ in the current node.
    *   $c$: Number of classes.
*   **Role in Decision Trees**: Similar to entropy, Decision Trees aim to reduce Gini impurity with each split. A lower Gini impurity indicates a more homogeneous node.

### Information Gain

*   **Definition**: Information Gain measures the reduction in entropy (or Gini impurity) achieved by splitting the data on a particular feature.
*   **Formula (using Entropy)**: $IG(S, A) = E(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} E(S_v)$
    *   $E(S)$: Entropy of the parent node $S$.
    *   $A$: The attribute (feature) being considered for the split.
    *   $Values(A)$: The possible values of attribute $A$.
    *   $S_v$: The subset of samples for which attribute $A$ has value $v$.
    *   $|S_v|/|S|$: The proportion of samples in $S_v$ relative to $S$.
*   **Role in Decision Trees**: At each step of building the tree, the algorithm chooses the feature that yields the highest Information Gain to make the split. This ensures that the chosen split maximizes the reduction in impurity, leading to more effective decision boundaries.
