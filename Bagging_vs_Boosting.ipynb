# Bagging vs Boosting

## Bagging (Bootstrap Aggregating)

### Definition
Bagging is an **ensemble learning technique** that improves model performance by training multiple models on different **random subsets** of the training data. The final prediction is made by combining the predictions of all models through **majority voting** (classification) or **averaging** (regression).

### Key Features

- Models are trained **independently and in parallel**.
- Uses **bootstrap sampling** (random sampling with replacement).
- Reduces **variance** and helps prevent overfitting.
- Commonly uses **Decision Trees** as base learners.

### Advantages

- Reduces overfitting.
- Improves prediction accuracy.
- Handles noisy data well.
- Training can be done in parallel.

### Disadvantages

- Requires more memory and computation.
- Less effective if the base models are very weak.
- Model interpretation becomes more difficult.

# Boosting

## Definition
Boosting is an **ensemble learning technique** that builds models **one after another**. Each new model focuses on correcting the mistakes made by the previous models. The final prediction is obtained by combining the predictions of all models using **weighted voting**.

### Key Features

- Models are trained **sequentially**.
- Gives more importance to previously misclassified samples.
- Reduces **bias** and improves accuracy.
- Common boosting algorithms include **AdaBoost**, **Gradient Boosting**, and **XGBoost**.

### Advantages

- High prediction accuracy.
- Works well with weak learners.
- Reduces bias.
- Produces strong predictive models.

### Disadvantages

- Sensitive to noisy data and outliers.
- Training is slower because models are built sequentially.
- More prone to overfitting if not tuned properly.

---

# Difference Between Bagging and Boosting

| Feature | Bagging | Boosting |
|---------|----------|----------|
| Training | Parallel | Sequential |
| Main Goal | Reduce Variance | Reduce Bias |
| Data Used | Random bootstrap samples | Entire dataset with updated sample weights |
| Model Dependency | Independent models | Each model depends on the previous one |
| Speed | Faster | Slower |
| Sensitivity to Noise | Low | High |
| Overfitting | Less likely | Can overfit if not tuned |
| Final Prediction | Majority voting / Averaging | Weighted voting |
| Example Algorithms | Random Forest | AdaBoost, Gradient Boosting, XGBoost |

---

# Summary

- **Bagging** trains multiple models independently on different subsets of data and combines their predictions to improve stability and reduce variance.
- **Boosting** trains models sequentially, with each model correcting the mistakes of the previous one to improve accuracy and reduce bias.
- **Random Forest** is a popular example of **Bagging**, while **AdaBoost** is a popular example of **Boosting**.
