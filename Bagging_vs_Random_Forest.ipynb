# Bagging vs Decision Tree

| Feature | Decision Tree | Bagging (Bootstrap Aggregating) |
|---------|---------------|---------------------------------|
| Definition | A single tree-based machine learning algorithm used for classification and regression. | An ensemble learning technique that combines multiple decision trees trained on different bootstrap samples. |
| Number of Models | One decision tree | Multiple decision trees |
| Training Data | Uses the entire dataset | Uses randomly sampled datasets (with replacement) |
| Accuracy | Moderate | Higher than a single decision tree |
| Overfitting | High chance of overfitting | Reduces overfitting by averaging predictions |
| Variance | High | Low |
| Bias | Low | Slightly higher but more stable |
| Prediction | Based on one tree | Majority voting (classification) or averaging (regression) |
| Robustness | Sensitive to noisy data | More robust to noise and outliers |
| Example | CART Decision Tree | Bagging Classifier, Random Forest |

## Advantages of Decision Tree
- Easy to understand and visualize.
- Requires little data preprocessing.
- Handles both numerical and categorical data.
- Fast for training and prediction.

## Disadvantages of Decision Tree
- Prone to overfitting.
- High variance.
- Small changes in data can produce different trees.

## Advantages of Bagging
- Reduces overfitting.
- Improves prediction accuracy.
- Produces more stable models.
- Works well with high-variance models like decision trees.

## Disadvantages of Bagging
- More computationally expensive.
- Requires more memory.
- Less interpretable than a single decision tree.

## Summary
- **Decision Tree:** A single, simple, and interpretable model but prone to overfitting.
- **Bagging:** Combines multiple decision trees to improve accuracy, stability, and generalization by reducing variance.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X,y = make_classification(n_features = 5, n_redundant = 0, n_informative = 5, random_state = 1)
```

```python
df = pd.DataFrame(X,columns = ['x1','x2','x3','x4','x5'])
df['y'] = y
print(df.shape)
df.head()
```

```python
bag = BaggingClassifier(max_features = 2)
```

```python
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python

plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
```

```python
rf = RandomForestClassifier(max_features=2)
```

```python
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python

plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
```
