# Naive Bayes Classifier

## Introduction
Naive Bayes is a **supervised machine learning algorithm** used for classification tasks. It is based on **Bayes' Theorem** and assumes that all features are conditionally independent given the class label. Despite this "naive" assumption, it performs well on many real-world problems such as spam detection, sentiment analysis, and document classification.

## Bayes' Theorem

The foundation of Naive Bayes is Bayes' Theorem:

$$
P(C \mid X) = \frac{P(X \mid C)\,P(C)}{P(X)}
$$

Where:

- $P(C \mid X)$ = Posterior probability of class $C$ given features $X$
- $P(X \mid C)$ = Likelihood of features $X$ given class $C$
- $P(C)$ = Prior probability of class $C$
- $P(X)$ = Evidence (probability of features $X$)

## Naive Bayes Formula

For multiple independent features $x_1, x_2, \ldots, x_n$:

$$
P(C \mid x_1, x_2, \ldots, x_n)
=
\frac{P(C)\prod_{i=1}^{n} P(x_i \mid C)}
{P(x_1, x_2, \ldots, x_n)}
$$

For classification, we choose the class with the highest posterior probability:

$$
\hat{C}
=
\arg\max_C
\left[
P(C)\prod_{i=1}^{n} P(x_i \mid C)
\right]
$$

## Advantages

- Simple and easy to implement
- Fast training and prediction
- Works well with high-dimensional data
- Effective for text classification problems

## Disadvantages

- Assumes feature independence, which may not hold in practice
- Performance can decrease when features are highly correlated
- Requires handling of zero probabilities (e.g., Laplace smoothing)

## Applications

- Spam Email Detection
- Sentiment Analysis
- Document Classification
- Medical Diagnosis
- Recommendation Systems

## Conclusion

Naive Bayes is a probabilistic classification algorithm based on Bayes' Theorem. Its simplicity, efficiency, and good performance on large datasets make it a popular choice for many classification tasks.
