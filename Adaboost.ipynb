# AdaBoost (Adaptive Boosting)

## Definition
AdaBoost (Adaptive Boosting) is an **ensemble learning algorithm** that combines multiple **weak learners** to create a **strong classifier**. It trains models one after another, where each new model focuses more on the samples that were incorrectly classified by the previous model. The most common weak learner used is a **Decision Stump** (a decision tree with one split).

---

## How AdaBoost Works

1. Assign equal weight to all training samples. Initially, for 'n' samples, $w_i = 1/n$.
2. Train the first weak learner $h_t(x)$.
3. Calculate the errors made by the model. The error rate for a weak learner $h_t$ is $\epsilon_t = \sum_{i=1}^n w_i I(y_i \neq h_t(x_i))$.
4. Increase the weights of the misclassified samples. The weight update rule is $w_i \leftarrow w_i \exp(-\alpha_t y_i h_t(x_i))$, where $\alpha_t = \frac{1}{2} \ln\left(\frac{1-\epsilon_t}{\epsilon_t}\right)$ is the weight of the weak learner.
5. Normalize the weights so they sum to 1.
6. Train the next weak learner using the updated weights.
7. Repeat the process for a fixed number of iterations $T$.
8. Combine all weak learners using weighted voting to make the final prediction: $H(x) = \text{sign}\left(\sum_{t=1}^T \alpha_t h_t(x)\right)$.

---

## Key Features

- Combines multiple weak learners into a strong model.
- Trains models sequentially.
- Gives more importance to misclassified samples.
- Commonly uses Decision Stumps as base learners.
- Improves prediction accuracy over a single model.

---

## Advantages

- High accuracy for classification tasks.
- Easy to implement.
- Works well with weak learners.
- Reduces classification errors.
- Less prone to overfitting than a single decision tree.

---

## Disadvantages

- Sensitive to noisy data and outliers.
- Training is slower because models are built one after another.
- Performance depends on the quality of the training data.

---

## Applications

- Spam Email Detection
- Face Detection
- Fraud Detection
- Medical Diagnosis
- Customer Churn Prediction

---

## Hyperparameters

| Hyperparameter | Description |
|---------------|-------------|
| `n_estimators` | Number of weak learners |
| `learning_rate` | Controls the contribution of each learner |
| `estimator` | Base model (usually a Decision Stump) |

---

## Time Complexity

- **Training:** $O(T \times N \times F)$
- **Prediction:** $O(T)$

Where:
- **T** = Number of weak learners
- **N** = Number of training samples
- **F** = Number of features

---

## Space Complexity

- **$O(N + T)$**

Where:
- **N** = Number of training samples
- **T** = Number of weak learners

---

## Summary

- AdaBoost stands for **Adaptive Boosting**.
- It is a **boosting ensemble algorithm**.
- It trains multiple weak learners sequentially.
- Each new learner focuses more on previously misclassified samples.
- The final prediction is made using **weighted voting** from all weak learners.
