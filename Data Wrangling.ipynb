```python
# Concept of DataFrame in Pandas
# From dictionary to DataFrame
# Most common way: pass dict of list/Series
# keys---> column name
# values----> column data
# index automatically created(0,1,2...)
# Transpose (.T): swaps the rows and columns--useful for reshaping 
```
