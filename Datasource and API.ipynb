## API
- Application Programmable Interface
- helps to communicate between the websites,server or devices
## Types:
### REST(Representational State Transfer)
- uses http method:
- GET = read data
- POST = create data
- PUT/PATCH = update
- DELETE = to delete
- returns JSON


```python
## REST DEMO
import requests
response = requests.get('https://pokeapi.co/api/v2/pokemon/pikachu')
data  = response.json()
print(f"Name : {data['name']}")
print(f"Types: {[t['type']['name'] for t in data['types']]}")
```

```python
import requests

BASE_URL = "https://jsonplaceholder.typicode.com/posts"

# 1. GET: Retrieve data (Read)
# Fetches the post with ID 1
get_response = requests.get(f"{BASE_URL}/1")
print("--- GET Response ---")
print(get_response.json()) 
```

## Create or Post



```python

new_post = {
    "title": "Learning REST APIs",
    "body": "This is a demonstration using Python Requests.",
    "userId": 1
}
post_response = requests.post(BASE_URL, json=new_post)
print("\n--- POST Response (Created) ---")
print(post_response.status_code) # 201 means "Created"
print(post_response.json())
```

## Update

```python
updated_data = {"title": "Updated Title", "body": "Updated content", "userId": 1}
put_response = requests.put(f"{BASE_URL}/1", json=updated_data)
print("\n--- PUT Response (Updated) ---")
print(put_response.json())
```

## Delete

```python
delete_response = requests.delete(f"{BASE_URL}/1")
print("\n--- DELETE Status Code ---")
print(delete_response.status_code) # 200 or 204 means success
```

## SOAP( Simple Object Access Protocol)
- Always XML envelopes (<soap:Envelope>)
- Uses WSDL (contract describing API)

```python
from zeep import Client

# The WSDL is the "contract" that defines what the API can do
wsdl_url = "https://www.dataaccess.com/webservicesserver/NumberConversion.wso?WSDL"

# Initialize the SOAP client
client = Client(wsdl=wsdl_url)

# Call the 'NumberToWords' service
# ubiNum is the parameter name defined in the WSDL
number = 5100000000000
result = client.service.NumberToWords(ubiNum=number)

print(f"The number {number} in words is: {result}")
```

```python
def safe_api_call(url):
    try:
        r = requests.get(url, timeout=10)
        if r.status_code == 200:
            return r.json()
        else:
            print(f"HTTP {r.status_code}")
    except Exception as e:
        print(f"Error: {e}")
    return None

# Test
safe_api_call('https://pokeapi.co/api/v2/pokemon/999999')

```
