### Decision Tree Hyperparameters Example

Decision Trees have several hyperparameters that can be tuned to control the complexity and performance of the model. Here's an example using `scikit-learn`.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Load a sample dataset (Iris dataset)
iris = load_iris()
X, y = iris.data, iris.target

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 1. Decision Tree with default hyperparameters
print("\n--- Default Decision Tree ---")
dtc_default = DecisionTreeClassifier(random_state=42)
dtc_default.fit(X_train, y_train)
y_pred_default = dtc_default.predict(X_test)
print(f"Accuracy (Default): {accuracy_score(y_test, y_pred_default):.4f}")

# 2. Decision Tree with custom hyperparameters
# Let's define some common hyperparameters:
#   - max_depth: The maximum depth of the tree. Controls overfitting.
#   - min_samples_split: The minimum number of samples required to split an internal node.
#   - criterion: The function to measure the quality of a split (e.g., 'gini' for Gini impurity, 'entropy' for information gain).
#   - random_state: Controls the randomness of the estimator.

print("\n--- Custom Hyperparameters Decision Tree ---")
dtc_custom = DecisionTreeClassifier(
    max_depth=3,                 # Limit the depth of the tree to prevent overfitting
    min_samples_split=10,        # Require at least 10 samples to split a node
    criterion='entropy',         # Use information gain for splitting
    random_state=42
)
dtc_custom.fit(X_train, y_train)
y_pred_custom = dtc_custom.predict(X_test)
print(f"Accuracy (Custom):  {accuracy_score(y_test, y_pred_custom):.4f}")

# You can also explore other hyperparameters like:
# - min_samples_leaf: The minimum number of samples required to be at a leaf node.
# - max_features: The number of features to consider when looking for the best split.
# - splitter: The strategy used to choose the split at each node ('best' or 'random').

print("\nKey Hyperparameters to Tune:")
print("  - max_depth: Controls model complexity, preventing overfitting.")
print("  - min_samples_split: Controls the minimum size of a node before it can be split.")
print("  - min_samples_leaf: Controls the minimum size of a leaf node.")
print("  - criterion: 'gini' (default) or 'entropy' for split quality.")
print("  - max_features: Number of features to consider for best split (useful for large datasets).")
```
