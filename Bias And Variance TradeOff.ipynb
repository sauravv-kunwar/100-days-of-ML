## Bias-Variance Trade-off

The Bias-Variance Trade-off describes the balance between a model's ability to learn patterns and its ability to generalize to new data.

### Total Error:

$$
\text{Total Error} = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}
$$

Where:

- **Bias**: Error due to overly simple assumptions
- **Variance**: Error due to sensitivity to training data
- **Irreducible Error**: Noise that cannot be removed

### Bias

High bias causes the model to miss important patterns.

$$
\text{High Bias} \Rightarrow \text{Underfitting}
$$

### Variance

High variance causes the model to fit noise in the training data.

$$
\text{High Variance} \Rightarrow \text{Overfitting}
$$

---

## Overfitting and Underfitting

### Underfitting

Underfitting occurs when the model is too simple to learn the underlying pattern of the data.

#### Characteristics:
- High training error
- High testing error
- Poor performance

#### Representation:

$$
Y \approx \beta_0 + \beta_1 X
$$

(Simple model)

---

### Overfitting

Overfitting occurs when the model learns both the pattern and noise in the training data.

#### Characteristics:
- Very low training error
- High testing error
- Poor generalization

#### Representation:

$$
Y = \beta_0 + \beta_1 X + \beta_2 X^2 + \cdots + \beta_n X^n
$$

(Excessively complex model)

---

## Ideal Model

The best model balances bias and variance for good generalization.

$$
\text{Optimal Model} \Rightarrow \text{Balanced Bias and Variance}
$$

### Key Points:
- High Bias → Underfitting
- High Variance → Overfitting
- Goal: Minimize total prediction error

```python
# Bias-Variance Trade-off
# Underfitting vs Overfitting Example

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Generate sample data
np.random.seed(0)

X = np.sort(5 * np.random.rand(80, 1), axis=0)
y = np.sin(X).ravel()

# Add noise
y = y + 0.2 * np.random.randn(80)

# Different polynomial degrees
degrees = [1, 4, 15]

# Create figure
plt.figure(figsize=(15, 5))

for i, degree in enumerate(degrees):

    # Create polynomial regression model
    model = make_pipeline(
        PolynomialFeatures(degree),
        LinearRegression()
    )

    # Train model
    model.fit(X, y)

    # Test data
    X_test = np.linspace(0, 5, 100).reshape(-1, 1)

    # Prediction
    y_pred = model.predict(X_test)

    # Plot graph
    plt.subplot(1, 3, i + 1)

    plt.scatter(X, y)
    plt.plot(X_test, y_pred)

    # Titles
    if degree == 1:
        plt.title("Underfitting")
    elif degree == 4:
        plt.title("Good Fit")
    else:
        plt.title("Overfitting")

    plt.xlabel("X")
    plt.ylabel("Y")

# Display plots
plt.tight_layout()
plt.show()
```
