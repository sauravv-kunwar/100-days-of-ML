# Softmax Regression

Softmax Regression is an extension of Logistic Regression used for **multi-class classification** problems where the target variable has more than two classes.

### Key Points
- Predicts probabilities for multiple classes.
- Uses the **Softmax Function** to convert scores into probabilities.
- The sum of all class probabilities equals 1.

### Softmax Formula

$$
P(y=i|x)=\frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}
$$

Where:
- $z_i$ = score (logit) for class $i$
- $K$ = total number of classes

---

# Training Intuition

The goal of training is to find the best weights and biases that minimize prediction errors.

### Steps
1. Initialize weights randomly.
2. Calculate class scores.
3. Apply Softmax to obtain probabilities.
4. Compute loss using Cross-Entropy.
5. Update weights using Gradient Descent.
6. Repeat until convergence.

### Score Calculation

$$
z = Wx + b
$$

Where:
- $W$ = weight matrix
- $x$ = input features
- $b$ = bias vector

---

# Loss Function

Softmax Regression uses **Categorical Cross-Entropy Loss**.

### Formula

$$
L = -\sum_{i=1}^{K} y_i \log(\hat{y}_i)
$$

Where:
- $y_i$ = actual class label (one-hot encoded)
- $\hat{y}_i$ = predicted probability

### For N Samples

$$
J = -\frac{1}{N}\sum_{n=1}^{N}\sum_{i=1}^{K}
y_{ni}\log(\hat{y}_{ni})
$$

### Purpose
- Penalizes incorrect predictions.
- Lower loss indicates better model performance.

---

# Prediction

After training, the model predicts the class with the highest probability.

### Prediction Rule

$$
\hat{y} = \arg\max_i P(y=i|x)
$$

### Example

If probabilities are:

$$
[0.10,\;0.75,\;0.15]
$$

Then:

$$
\hat{y}=2
$$

because 0.75 is the highest probability.

---

# Sigmoid vs Softmax

| Feature | Sigmoid | Softmax |
|----------|----------|----------|
| Output Range | 0 to 1 | 0 to 1 |
| Number of Outputs | One | Multiple |
| Sum of Probabilities | Not necessarily 1 | Always 1 |
| Use Case | Binary Classification | Multi-Class Classification |
| Formula | $\sigma(z)=\frac{1}{1+e^{-z}}$ | $\frac{e^{z_i}}{\sum e^{z_j}}$ |

### Sigmoid Formula

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

### Softmax Formula

$$
P(y=i|x)=\frac{e^{z_i}}{\sum_{j=1}^{K}e^{z_j}}
$$



# Summary

- **Softmax Regression** is used for multi-class classification.
- It converts class scores into probabilities using the Softmax function.
- Training minimizes **Cross-Entropy Loss**.
- Prediction is made using the **argmax** of probabilities.
- **Sigmoid** is for binary classification, while **Softmax** is for multi-class classification.

```python
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
 df.head()
```

```python
encoder = LabelEncoder()
df['species'] = encoder.fit_transform(df['species'])
```

```python
df.head()
```

```python
df = df[['sepal_length','petal_length','species']]
```

```python
df.head()
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
clf = LogisticRegression( multi_class = 'multinomial')
```

```python
clf.fit(X_train,y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
pd.DataFrame(confusion_matrix(y_test,y_pred))
```

```python
# prediction
query = np.array([[3.4,2.7]])
clf.predict_proba(query)
```
