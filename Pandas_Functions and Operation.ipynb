```python
frame  = pd.DataFrame(np.random.standard_normal((4,3)),
                      columns = list('bde'),
                      index = ['Utah','Ohio','Texas','Oregon'])
frame
```

```python
np.abs(frame)
```

```python
def f1(x):
    return x.max() - x.min()
    
```

```python
frame.apply(f1)
```

```python
frame.apply(f1,axis = 'columns')
```

```python
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])
frame.apply(f2)
```

```python
def my_format(x):
    return f"{x:.2f}"
frame.map(my_format)


# apply() ---> applies a function column-wise or row-wise
# map() ----> applies a function to each individual elements
```
